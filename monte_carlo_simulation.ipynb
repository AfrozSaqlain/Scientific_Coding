# Estimating the value of $\pi$

Imagine there's a square with side length equal to 2 units and a circle inscribed inside it 

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

# Create a figure and axes
fig, ax = plt.subplots()

# Square: bottom-left corner (0, 0), side length 2
square = Rectangle((0, 0), 2, 2, fill=False, edgecolor="blue", linewidth=2)

# Circle: center (1, 1), radius 1
circle = Circle((1, 1), 1, fill=False, edgecolor="red", linewidth=2)

# Add shapes to the plot
ax.add_patch(square)
ax.add_patch(circle)

# Set plot boundaries and equal scaling
ax.set_xlim(-0.2, 2.2)
ax.set_ylim(-0.2, 2.2)
ax.set_aspect("equal")

plt.show()
```
