# Estimating the value of $\pi$

Imagine there's a square with side length equal to 2 units and a circle inscribed inside it 

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

fig, ax = plt.subplots()

square = Rectangle((-1, -1), 2, 2, fill=False, edgecolor="blue", linewidth=1)

circle = Circle((0, 0), 1, fill=False, edgecolor="red", linewidth=1)

ax.add_patch(square)
ax.add_patch(circle)

ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.set_aspect("equal")

plt.show()
```

If we only take one-fourth of this figure, we will have a square of area 1 units and a sector of a circle with area $\pi / 4$.

Now if we try to select random points form that part of the above figure, the probability of a point being inside the circle's sector $ = \frac{\pi / 4}{1} = \frac{\pi}{4}$

Repeating the process of sampling multiple points we should be able to converge the empirical probability to theoretical probability, by the Law of large numbers.

```python
import numpy as np
```

```python
trials = 100000
count = 0

for i in range(trials):
    x = np.random.uniform(0, 1)
    y = np.random.uniform(0, 1)

    if x**2 + y**2 <= 1:
        count += 1

print(f"Estimated value of π: {4 * count / trials}")
```
